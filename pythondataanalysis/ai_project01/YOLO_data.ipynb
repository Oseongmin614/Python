```python
import os
import glob
import shutil
import random
from collections import defaultdict
```

```python
orig_img_dir = r"C:\ai_project01\path_guide_dataset\dataset02\images"
orig_lbl_dir = r"C:\ai_project01\path_guide_dataset\dataset02\labels"
aug_img_dir = r"C:\ai_project01\path_guide_dataset\dataset02\augments\images"
aug_lbl_dir = r"C:\ai_project01\path_guide_dataset\dataset02\augments\labels"
```

```python
base_dst=r"C:\ai_project01\path_guide_dataset\dataset03"
```

```python
dst_img_train = os.path.join(base_dst,"images","train")
dst_img_val = os.path.join(base_dst,"images","val")
dst_lbl_train = os.path.join(base_dst,"labels","train")
dst_lbl_val = os.path.join(base_dst,"labels","val")
```

```python
class_list = [
    'person', 'bicycle', 'car', 'motorcycle', 'bus', 'truck',
    'movable_signage', 'bollard', 'fire_hydrant', 'bench',
    'chair', 'table', 'tree_trunk', 'potted_plant',
    'traffic_light', 'traffic_sign'
]
```

```python
for folder in [dst_img_train,dst_img_val,dst_lbl_train,dst_lbl_val]:
    os.makedirs(folder,exist_ok = True)
    
```

```python
orig_images = glob.glob(os.path.join(orig_img_dir,"*.png"))
aug_images=glob.glob(os.path.join(aug_img_dir,"*.png"))

all_images=orig_images+aug_images
random.shuffle(all_images)

split_idx = int(len(all_images)*0.8)
train_imgs = all_images[:split_idx]
val_imgs= all_images[split_idx:]
```

```python
def is_valid_label_file(label_path):
    try:
        with open(label_path,"r")as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) != 5:
                    return False
                class_id = float(parts[0])
                if not class_id.is_integer():
                   return False
                class_id = int(class_id)
                if class_id<0 or class_id>=len(class_list):
                    return False
                cx,cy,w,h = map(float,parts[1:])
                if not all(0.0 <= v <= 1.0 for v in [cx,cy,w,h]):
                    return False
            return True
    except Exception as e:
        print(f"!!라벨오류  ->   {label_path}:{e}")
        return False
        
```

```python
def copy_pair(image_path, src_img_dir,src_lbl_dir,dst_img_dir,dst_lbl_dir):
    base = os.path.splitext(os.path.basename(image_path))[0]
    label_path = os. path.join(src_lbl_dir, base + ".txt")

    if os.path.exists(label_path):
        if not is_valid_label_file(label_path):
            print(f"!!잘못된 라벨   -> {label_path}")
            return
    shutil.copy2(image_path,os.path.join(dst_img_dir,os.path.basename(image_path)))
    if os.path.exists(label_path):
        shutil.copy2(label_path, os.path.join(dst_lbl_dir,os.path.basename(label_path)))
```

```python
for img_path in train_imgs:
    src_img_dir = aug_img_dir if 'aug' in img_path else orig_img_dir
    src_lbl_dir = aug_lbl_dir if 'aug' in img_path else orig_lbl_dir
    copy_pair(img_path,src_img_dir,src_lbl_dir,dst_img_train,dst_lbl_train)
    
```

```python
for img_path in val_imgs:
    src_img_dir=aug_img_dir if 'aug' in img_path else orig_img_dir
    src_lbl_dir = aug_lbl_dir if 'aug' in img_path else orig_lbl_dir
    copy_pair(img_path, src_img_dir,src_lbl_dir, dst_img_val,dst_lbl_val)
```
