```python
import os
from ultralytics import YOLO
```

```python
dataset_dir = r'C:\ai_project01\path_guide_dataset\dataset03'
yaml_path= os.path.join(dataset_dir,'data.yaml')
```

```python
class_list = [
    'person', 'bicycle', 'car', 'motorcycle', 'bus', 'truck',
    'movable_signage', 'bollard', 'fire_hydrant', 'bench',
    'chair', 'table', 'tree_trunk', 'potted_plant',
    'traffic_light', 'traffic_sign'
]
```

```python
yaml_content = f"""path:{dataset_dir.replace(os.sep,'/')}
train: images/train
val:images/val
nc:{len(class_list)}
names:{class_list}
"""
```

```python
with open(yaml_path,'w',encoding ='utf-8')as f:
    f.write(yaml_content)

print(f"data.yaml 저장 완료 >> {yaml_path}")
```

```python
model = YOLO('yolov8n.pt')
```

```python
model.train(
    data=yaml_path,
    epochs=3,
    imgsz=640,
    batch=64,
    workers=8,
    device=0,
    cache=False,
    project=r'C:\ai_project01\yolo_train_results',
    name='yolov8n_path_guide',
    exist_ok=True
)
print("YOLOv8 학습완료")
```
