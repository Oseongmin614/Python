```python
import cv2
from ultralytics import YOLO
```

```python
cap =cv2.VideoCapture('walk.mp4')
```

```python
if cap.isOpened() == False:
    print("영상을 열지 못했습니다")
    exit()
```

```python
yolo_model = YOLO(r'C:\ai_project01\yolo_train_results\yolov8n_path_guide\weights\best.pt')
```

```python
while True:
    ret, frame = cap.read()
    if ret ==False:
        break
    results = yolo_model(frame, verbose=False)[0]
    frame = results.plot()
    cv2.imshow("GuidePath",frame)
    key = cv2.waitKey(50)&0xFF
    if key== ord('q'):
        break
```

```python
cap.release()
cv2.destroyAllWindows()
```
