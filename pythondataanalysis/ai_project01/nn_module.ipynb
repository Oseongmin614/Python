```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
h=torch.tensor([1,0,0,0],dtype=torch.float32)
e=torch.tensor([0,1,0,0],dtype=torch.float32)
l=torch.tensor([1,0,1,0],dtype=torch.float32)
o=torch.tensor([0,0,0,1],dtype=torch.float32)
```

```python
inputs = torch.stack([h,e,l,l],dim = 0)
target = torch.tensor([3])
```

```python
class TinyCharRNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.Wxh = nn.Parameter(torch.tensor([0.5, -0.3, 0.8, -0.2], dtype=torch.float32))
        self.b = nn.Parameter(torch.tensor(0.0, dtype=torch.float32))
        self.Whh = nn.Parameter(torch.tensor(0.4, dtype=torch.float32))
        self.Why = nn.Parameter(torch.tensor([0.6, -0.3, 0.2, 0.8], dtype=torch.float32))

    def forward(self, x_seq: torch.Tensor) -> torch.Tensor:
        h_t = x_seq.new_tensor(0.0)
        for x_t in x_seq:
            Wxh_x = torch.dot(x_t, self.Wxh) + self.b
            h_prev = h_t * self.Whh
            h_t = torch.tanh(Wxh_x + h_prev)
        y_raw = self.Why * h_t
        return y_raw.view(1, -1)
```

```python
model = TinyCharRNN()
```

```python
optimizer = optim.SGD(model.parameters(),lr=0.1)
```

```python
for epoch in range(100):
    logits = model(inputs)
    loss = F.cross_entropy(logits, target)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if (epoch+1)%1==0:
        with torch.no_grad():
            pred_prob = F.softmax(model(inputs),dim=1)
            pred_idx= int(torch.argmax(pred_prob))
        print("="*100)
        print("업데이트되 가중치")
        print(f"Wxh={model.Wxh.detach().cpu().numpy()}")
        print(f"b={model.b.item():.6f}")
        print(f"Whh={model.Whh.item():.6f}")
        print(f"Why={model.Why.detach().cpu().numpy()}")
        print("="*100)
        
        print(f"에폭:{epoch+1:3d}|loss:{loss.item():.4f}")
        print(f"예측확률:{pred_prob.cpu().numpy()}")
        print(f"예측 인덱스:{pred_idx}")
        print(f"예측글자:{['h','e','l','o'][pred_idx]}")
```
