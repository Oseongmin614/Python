```python
import cv2
import matplotlib.pyplot as plt
```

```python
video_path = 'C:/ai_project01/incredibles.mp4'
cap = cv2.VideoCapture(video_path)
```

```python
frame_inter = 15
count = 0
displayed_frames=0
```

```python
plt.figure(figsize=(30,20))
while cap.isOpened() and displayed_frames <10:
    ret,frame = cap.read()
    if not ret:
        break
    if count % frame_inter ==0:
        plt.subplot(5,2,displayed_frames +1)
        plt.imshow(frame)
        plt.title(f'Frame{count}')
        displayed_frames +=1
    count+=1
cap.release()
plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
```

```python
video_path = 'C:/ai_project01/incredibles.mp4'
cap = cv2.VideoCapture(video_path)
```

```python
frame_interval= 10
count=0 
displayed_frames =0
```

```python
model = YOLO('C:/ai_project01/incredible_yolo_model.pt')
```

```python
plt.figure(figsize=(30,20))
while cap.isOpened() and displayed_frames < 10:
    ret, frame = cap.read()
    if not ret:
        break
    if count %frame_interval ==0:
        plt.subplot(5,2,displayed_frames+1)
        results = model(frame, conf=0.5)
        annotated_frame= results[0].plot()
        plt.imshow(annotated_frame)
        plt.title(f'Frame{count}')
        displayed_frames +=1
    count +=1
cap.release()
plt.tight_layout()
plt.show()
```
