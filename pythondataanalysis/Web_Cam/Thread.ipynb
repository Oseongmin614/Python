```python
import threading 
import time 
```

```python
def go():
    for i in range(1,101):
        print(f"\n{i}번째:go")
        time.sleep(0.1)

def come():
    for i in range(1,101):
        print(f"\n{i}번째 : come")
        time.sleep(0.1)
```

```python
go()
come()
```

```python
t1= threading.Thread(target=go)
t2 = threading.Thread(target=come)
t1.start()
t2.start()
t1.join()
t2.join()
```
